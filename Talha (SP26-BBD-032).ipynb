```python
#Q3
n = int(input("Enter a positive integer: "))

i = 1
total = 0

while i <= n:
    total += i
    i += 1

print("Sum from 1 to", n, "is:", total)
```

```python
#Q7
num = int(input("Enter a number: "))

count = 0
if num == 0:
    count = 1
else:
    num = abs(num)  
    while num > 0:
        num //= 10
        count += 1

print("Number of digits:", count)
```

```python
#Q8
total = 0
num = int(input("Enter a number (0 to stop): "))

while num != 0:
    total += num
    num = int(input("Enter a number (0 to stop): "))

print("Final total:", total)
```
