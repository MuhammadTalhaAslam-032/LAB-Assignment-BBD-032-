```python
# 1. Simple Function
from datetime import datetime

def greet_with_time(name):
    current_time = datetime.now().strftime("%H:%M:%S")
    print(f"Hello, {name}! Current time is {current_time}")


greet_with_time("Ali")
```

```python
# 2. Function with Multiple Parameters
def calculate_average(num1, num2, num3):
    return (num1 + num2 + num3) / 3


avg = calculate_average(10, 20, 30)
print("Average:", avg)
```

```python
def print_info(name, age, city):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")


# Calling the function using keyword arguments
print_info(name="Ali", age=20, city="Faisalabad")
```

```python
# 3. Default Argument
def calculate_discount(price, discount_percentage=10):
    discount = price * (discount_percentage / 100)
    final_price = price - discount
    return final_price
print("Discounted Price:", calculate_discount(1000))
print("Discounted Price:", calculate_discount(1000, 20))
```

```python
# 5. Variable-Length Arguments
def find_max(*numbers):
    return max(numbers)


print("Maximum:", find_max(3, 7, 2, 9, 15))
```

```python
# 6. Recursive Function
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


print("Factorial:", factorial(5))
```

```python
# 7. Function as an Argument
def apply_function(func, numbers):
    return [func(num) for num in numbers]


def double(x):
    return x * 2


print("Applied Function:", apply_function(double, [1, 2, 3, 4]))
```

```python
# 9. Higher-Order Function
def apply_operation(func, numbers):
    return [func(num) for num in numbers]


# Define square function
def square(x):
    return x ** 2


print("Squared List:", apply_operation(square, [1, 2, 3, 4]))
```

```python
# 10. Function Decorator
import time

def timer_decorator(func):
    def wrapper():
        start_time = time.time()
        
        func()
        
        end_time = time.time()
        print(f"Execution Time: {end_time - start_time:.6f} seconds")
    
    return wrapper


@timer_decorator
def sample_function():
    time.sleep(1)
    print("Function is running...")


sample_function()
```
