```python
# D
orders = [
    {"product": "phone", "category": "electronics"},
    {"product": "shirt", "category": "clothing"},
    {"product": "laptop", "category": "electronics"},
    {"product": "jeans", "category": "clothing"},
    {"product": "tv", "category": "electronics"}
]

category_count = {}

for order in orders:
    cat = order["category"]
    category_count[cat] = category_count.get(cat, 0) + 1

most_popular = max(category_count, key=category_count.get)

print("Most popular category:", most_popular)
```

```python
# E
# Create a dictionary with student names and their grades
student_grades = {
    "Ali": 85,
    "Sara": 90,
    "Ahmed": 78,
    "Zainab": 92
}

# Calculate the average grade
total = sum(student_grades.values())
number_of_students = len(student_grades)
average_grade = total / number_of_students

# Print results
print("Student Grades:", student_grades)
print("Average Grade:", average_grade)
```

```python
# F
# Inventory dictionary: name -> [quantity, price]
inventory = {
    "Pen": [10, 20],
    "Book": [5, 100],
    "Pencil": [2, 10]
}

# Add new product
inventory["Eraser"] = [8, 15]

# Update quantity
inventory["Pencil"][0] += 3

# Calculate total value
total = 0
for item in inventory:
    qty, price = inventory[item]
    total += qty * price

print("Total Inventory Value:", total)

# Low stock items (less than or equal to 5)
print("Low Stock Items:")
for item in inventory:
    if inventory[item][0] <= 5:
        print(item, inventory[item][0])
```

```python
# G
# name -> [Math, English, Science]
students = {
    "Ali": [85, 90, 88],
    "Sara": [92, 95, 91],
    "Ahmed": [70, 75, 72]
}

# Average of each student
for name in students:
    avg = sum(students[name]) / 3
    print(name, "Average:", avg)

# Highest and lowest in Math
math_marks = []
for marks in students.values():
    math_marks.append(marks[0])

print("Highest Math:", max(math_marks))
print("Lowest Math:", min(math_marks))

# Sort by average
print("Sorted by Average:")
for name in sorted(students, key=lambda x: sum(students[x])/3, reverse=True):
    print(name, sum(students[name]) / 3)
```

```python
# H
# Store: name -> [price, quantity]
store = {
    "Pen": [20, 10],
    "Book": [100, 5],
    "Pencil": [10, 20]
}

# Cart: name -> quantity
cart = {}

# Add to cart
cart["Pen"] = 2
cart["Book"] = 1

# Remove from cart
del cart["Book"]

# Calculate total
total = 0
for item in cart:
    price = store[item][0]
    qty = cart[item]
    total += price * qty

print("Total:", total)

# Apply discount (10%)
discount = total * 0.10
total_after_discount = total - discount

# Apply tax (5%)
tax = total_after_discount * 0.05
final_total = total_after_discount + tax

print("Final Total:", final_total)
```
